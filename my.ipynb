```python
%cd "F:/Research/AAAI Computer vision/GIFT_3D_Representation_Study"

import matplotlib.pyplot as plt
from repstudy.datasets import NYUv2Adapter

adapter = NYUv2Adapter(
    root="data/nyu",
    split="test",
    intrinsics=[518.857901, 519.469611, 325.582449, 253.736166],
)

sample_id = adapter.sample_ids()[0]
sample = adapter.load(sample_id)

fig, axes = plt.subplots(1, 4, figsize=(18, 5))

axes[0].imshow(sample.rgb)
axes[0].set_title(f"RGB: {sample_id}")

axes[1].imshow(sample.depth_m, cmap="magma")
axes[1].set_title("Depth in metres")

axes[2].imshow(sample.semantic_gt, cmap="tab20")
axes[2].set_title("Semantic labels")

axes[3].imshow(sample.valid_depth_mask, cmap="gray")
axes[3].set_title("Valid depth mask")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()

print("Image shape:", sample.rgb.shape)
print("Depth range:", sample.depth_m.min(), "to", sample.depth_m.max(), "metres")
print("Valid depth pixels:", sample.valid_depth_mask.sum())
```
