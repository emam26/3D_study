# catDog image and OBJ representation grids

The RGB scene and the standalone Cat OBJ asset are shown together. The OBJ-derived representations are true 3D; they are not camera-aligned reconstructions of the RGB scene.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

if not (Path.cwd() / "repstudy").exists():
    raise RuntimeError("Open this notebook with the study repository as the Jupyter working directory")

paths = [
    Path("outputs/catdog/catdog_3d_representation_grid.png"),
    Path("outputs/catdog/catdog_2d_representation_grid.png"),
]
fig, axes = plt.subplots(2, 1, figsize=(22, 18))
for axis, path in zip(axes, paths):
    axis.imshow(Image.open(path))
    axis.set_title(path.name)
    axis.axis("off")
plt.tight_layout()
plt.show()
```
