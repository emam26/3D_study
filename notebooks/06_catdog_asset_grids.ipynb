# Cat3D OBJ representation atlas

The 3D atlas starts with the original Cat OBJ/CAD model and then shows its nine derived representations. The 2D atlas shows the same representations projected into the front view.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

if not (Path.cwd() / "repstudy").exists():
    raise RuntimeError("Open this notebook with the study repository as the Jupyter working directory")

paths = [
    Path("outputs/cat3d/cat3d_original_plus_9_grid.png"),
    Path("outputs/cat3d/cat3d_2d_representation_grid.png"),
]
fig, axes = plt.subplots(2, 1, figsize=(24, 20), squeeze=False)
for axis, path in zip(axes.flat, paths):
    axis.imshow(Image.open(path))
    axis.set_title(path.name)
    axis.axis("off")
plt.tight_layout()
plt.show()
```
