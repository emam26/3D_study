# Cat3D OBJ representation atlas

The first panel is the original Cat OBJ/CAD model; the remaining panels are its nine derived 3D representations.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

if not (Path.cwd() / "repstudy").exists():
    raise RuntimeError("Open this notebook with the study repository as the Jupyter working directory")

paths = [Path("outputs/cat3d/cat3d_original_plus_9_grid.png")]
fig, axes = plt.subplots(1, 1, figsize=(24, 11), squeeze=False)
for axis, path in zip(axes.flat, paths):
    axis.imshow(Image.open(path))
    axis.set_title(path.name)
    axis.axis("off")
plt.tight_layout()
plt.show()
```
