# Multiview projection grids

Each row compares the original camera, left-oblique, right-oblique, and elevated projections for a point cloud, graph, octree, or local descriptor representation.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

if not (Path.cwd() / "repstudy").exists():
    raise RuntimeError("Open this notebook with the study repository as the Jupyter working directory")

paths = [
    Path("outputs/nyuv2/nyuv2_representation_smoke/projection_grid.png"),
    Path("outputs/sunrgbd/sunrgbd_representation_smoke/projection_grid.png"),
]
fig, axes = plt.subplots(2, 1, figsize=(22, 30))
for axis, path in zip(axes, paths):
    axis.imshow(Image.open(path))
    axis.set_title(path.parts[-3])
    axis.axis("off")
plt.tight_layout()
plt.show()
```
