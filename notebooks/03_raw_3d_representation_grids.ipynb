# Raw 3D representation grids

These are the two generated 2x4 figures. Each includes RGB, semantic ground truth, and the six raw 3D geometries—not the 2D coverage or oracle error map.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

if not (Path.cwd() / "repstudy").exists():
    raise RuntimeError("Open this notebook with the study repository as the Jupyter working directory")

paths = [
    Path("outputs/nyuv2/nyuv2_representation_smoke/3d_representation_grid.png"),
    Path("outputs/sunrgbd/sunrgbd_representation_smoke/3d_representation_grid.png"),
]
fig, axes = plt.subplots(2, 1, figsize=(18, 20))
for axis, path in zip(axes, paths):
    axis.imshow(Image.open(path))
    axis.set_title(path.parts[-3])
    axis.axis("off")
plt.tight_layout()
plt.show()
```
