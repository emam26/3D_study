# 2D representation-projection grids

Each representation is projected back into the camera view and shown beside RGB and semantic ground truth. Coverage is reported in each title.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

if not (Path.cwd() / "repstudy").exists():
    raise RuntimeError("Open this notebook with the study repository as the Jupyter working directory")

paths = [
    Path("outputs/nyuv2/nyuv2_representation_smoke/2d_representation_grid.png"),
    Path("outputs/sunrgbd/sunrgbd_representation_smoke/2d_representation_grid.png"),
]
fig, axes = plt.subplots(2, 1, figsize=(22, 20))
for axis, path in zip(axes, paths):
    axis.imshow(Image.open(path))
    axis.set_title(path.parts[-3])
    axis.axis("off")
plt.tight_layout()
plt.show()
```
